```python
class Person:
    def __init__(self, name, age, city, country):
        self.name = name
        self.age = age
        self.city = city
        self.country = country

p1 = Person("Kajal", 26, "Banglore", "India")

print(p1.name)
print(p1.age)
print(p1.city)
print(p1.country)
```

```python
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

p1 = Person("Kajal", 26)

print(p1.name)
print(p1.age)
```

```python
class Person:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print("Hello, my name is " + self.name)

p1 = Person("Kajal") 
p1.greet()       
```

```python
class Calculator:
    def add(self, a, b):
        return a + b
    
    def multiply(self, a, b):
        return a * b
    
calc = Calculator() 
print(calc.add(5, 4))  
print(calc.multiply(5, 4)) 
```
